**Pandas Introduction**

Pandas is an open-source Python library used for data manipulation, analysis and cleaning. It provides fast and flexible tools to work with tabular data, similar to spreadsheets or SQL tables

```python
import pandas as pd
```

**Pandas Series**




A Pandas Series is a one-dimensional labeled array capable of holding data of any type

```python
data=[1,2,3,4]
s=pd.Series(data)
print('Pandas Series:' )
print(s)
```

**Pandas Dataframe**


Pandas DataFrame is a two-dimensional data structure with labeled axes (rows and columns).


```python
import pandas as pd

df = pd.DataFrame()
print(df)
lst = ['Geeks', 'For', 'Geeks', 'is', 'portal', 'for', 'Geeks']

df = pd.DataFrame(lst)
print(df)
```
