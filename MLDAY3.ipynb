```python
from sklearn.metrics import mean_absolute_error, mean_squared_error ,root_mean_squared_error ,r2_score
```

```python
y_original=[10,20,30]
y_predicted=[12,18,31]

mae=mean_absolute_error(y_original,y_predicted)
print(mae)
```

```python
y_original=[10,20,30]
y_predicted=[12,18,31]

mse=mean_squared_error(y_original,y_predicted)
print(mse)
```

```python
y_original=[10,20,30]
y_predicted=[12,18,31]

r=root_mean_squared_error(y_original,y_predicted)
print(r)
```

```python
y_original=[10,20,30]
y_predicted=[12,18,31]

r2=r2_score(y_original,y_predicted)
print(r2)
```

LINEAR REGRESSION


```python
import numpy as np
from sklearn.linear_model import LinearRegression

x=np.array([[1],[2],[3],[4],[5]])
y=np.array([35,40,45,52,60])

model=LinearRegression()

model.fit(x,y)

prid=model.predict([[6],[7],[8]])
print(prid)
```

```python
import matplotlib.pyplot as plt

plt.plot(x,y)
plt.title('Linear Regression')
plt.xlabel('Number of Hours')
plt.ylabel('Marks Obtained')
plt.show()
```
