**WHAT IS DATA ANALYSIS??**

Data Analysis is a process of inspecting,cleansing,transforming and modeling data with the goal of discovering useful information,informing conclusions and supporting decision-making.

**Parts of DAP**

* ASKING QUESTIONS
* DATA WRANGLING (PROCESSING,CLEANING)
* EDA
* DRAWING CONCLUSIONS
* COMMUNICATING RESULTS

**Data Wrangling**

Data wrangling, sometimes refers to as data munging ,is the process of transforming and mapping data from one 'raw' data form into another format with the intent of making it more appropriate and valuable for a variety of downstream purposes such as analytics.

**STEPS OF DATA WRANGLING**
* Gathering data
* Assessing Data
* Cleaning Data
