<h2>Simple Linear Regression</h2>

<h3>Import all the libraries</h3>

```python
import numpy as np    #mathematical operations
import matplotlib.pyplot as plt    #nice graphs
from mpl_toolkits.mplot3d import Axes3D  #3d animation
```

<h3>Generate input data to train on</h3>

```python
observations=1000
xs=np.random.uniform(low=-10,high= 10,size=(observations,1))
zs=np.random.uniform(-10,10,(observations,1))

inputs=np.column_stack((xs,zs))
print(inputs.shape)
```

<h3>Create the targets we will aim at</h3>

```python
noise=np.random.uniform(-1,1,(observations,1))

targets= 2*xs - 3*zs + 5 + noise
print(targets.shape)
```

<h3>Plot the data</h3>

```python
targets=targets.reshape(observations,)
fig=plt.figure()
ax=fig.add_subplot(111,projection='3d')
ax.plot(xs,zs,targets)
ax.set_xlabel('xs')
ax.set_ylabel('zs')
ax.set_zlabel('Targets')
ax.view_init(azim=100)
plt.show()
targets=targets.reshape(observations,1)
```

<h3>Initialize Variables</h3>

```python
init_range=0.1

weights= np.random.uniform(-init_range,init_range,size=(2,1))
biases= np.random.uniform(-init_range,init_range,size=1)
print(weights)
print(biases)
```

<h3>Set a learning rate (ETA RATE)</h3>

```python
learning_rate=0.02
```

TRAIN THE MODEL

```python
for i in range(100):
    out= np.dot(inputs,weights) +biases
    deltas=out- targets
    loss=np.sum(deltas**2)/2 /observations
    print(loss)

    deltas_scaled= deltas/observations
    weights=weights -learning_rate*np.dot(inputs.T,deltas_scaled)
    biases= biases -learning_rate*np.sum(deltas_scaled)
```

PRINT THE WEIGHTS AND BIASES AND SEE IF WE HAVE WORKED CORRECTLY

```python
print(weights,biases)
```

PLOT LAST OUTPUTS VS TARGETS

```python
plt.plot(out,targets)
plt.show()
```
