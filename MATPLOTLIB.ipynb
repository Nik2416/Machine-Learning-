```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
from matplotlib.lines import lineStyles
x=[1,2,3,4]
y=[10,20,30,40]

plt.plot(x,y,'-r',ms =13)
plt.title('Line plot')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.grid(True)
```

```python
x=['maths','science','english','geography','hindi','history']
y=[90,89,79,75,95,88]
plt.bar(x,y,color='violet')
plt.title('Bar Plot')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.show()
```

```python
x=[1,2,3,4]
y=[10,20,30,40]
plt.scatter(x,y)
plt.title('Scatter Plot')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
```

```python
x=[11,22,33,44]

plt.hist(x)
plt.title('Histogram Plot')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
```

```python
x=['Maths ','Hindi ','History','English']
y=[90,80,70,60]
plt.pie(y,labels=x)
plt.title('Pie Plot')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
```

```python
import numpy as np
x=[2,4,6,8,10]
y=[1,2,3,4,5]
x1=np.array(y)
y1=x1*2
plt.plot(x1,y1,label='y=2x')
plt.plot(x,y,label='y=x')
plt.title('Line Plot')
plt.fill_between(x,y,x1,y1,color='blue')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.legend()
```

```python
import numpy as np
import matplotlib.pyplot as plt
x=np.linspace(0,2*np.pi,100)
y_sin=np.sin(x)
y_cos=np.cos(x)

plt.plot(x,y_sin,label='sin(x)')
plt.plot(x,y_cos,label='cos(x)')
plt.legend()
```
