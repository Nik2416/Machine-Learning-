**MULTIPLE** **REGRESSION**

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
```

```python
df=pd.read_csv('housing.csv')
```

```python
df_copy=df.copy()
```

```python
df_copy.dropna(inplace=True)
```

```python
df_copy.isnull().sum()
```

```python
df_copy=df_copy.drop(['population','households'],axis=1)
```

```python
df_copy=df_copy.drop(['ocean_proximity'],axis=1)
```

```python
df_copy
```

```python
df_copy=df_copy.drop(['housing_median_age'],axis=1)
```

```python
x=df_copy.drop(['median_house_value'],axis=1)
y=df_copy['median_house_value']
```

```python
x
```

```python
# Split data into training and testing sets
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)
```

```python
x_train
```

```python
y_train
```

```python
model=LinearRegression()
model.fit(x_train,y_train)
```

```python
# Make predictions
y_pred = model.predict(x_test)

# Model coefficients
print("Intercept:", model.intercept_)
print("Coefficients:", model.coef_)

# Model evaluation
print("R² Score:", r2_score(y_test, y_pred))
print("Mean Squared Error:", mean_squared_error(y_test, y_pred))

```
